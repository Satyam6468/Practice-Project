```python
def sleepBins(x):
    if(x<3):
        return "<3"
    elif(x<6):
        return "3-6"
    elif(x<9):
        return "6-9"
    else:
        return "9+"
```

```python
df["Sleep_range"]=df["sleep_hours"].apply(sleepBins)
```
